# Find the unique number
There is an array with some numbers. All numbers are equal except for one. Try to find it!

```python
def find_uniq(xs):
    for num in set(xs):
        if xs.count(num) < 2:
            return num
```

### EJEMPLOS

```python
find_uniq([ 1, 1, 1, 2, 1, 1 ])
```

```python
find_uniq([5,5,5,5,5,5,5,5,5.055,5,5])
```
